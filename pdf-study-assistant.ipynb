```python
# CÉLULA 1 — Importar as funções e modelos do projeto
from pdf_study import (
    ResultadoAnalise,
    TopicoEstudo,
    analisar_documento,
    extrair_texto_pdf,
    validar_topicos,
)
```

```python
# CÉLULA 2 — Extrair texto do PDF (sem chamar Gemini)
document, total_paginas = extrair_texto_pdf("introducao_pyhton.pdf")
print(f"Páginas totais: {total_paginas}")
print(f"Páginas com texto: {len(document)}")
```

```python
# CÉLULA 3 — Testar a validação local, sem gastar API
exemplo = ResultadoAnalise(topicos=[
    TopicoEstudo(nome="Exemplo válido", explicacao="Teste", importancia="Teste", pagina=4),
    TopicoEstudo(nome="Página inválida", explicacao="Teste", importancia="Teste", pagina=35),
])
for item in validar_topicos(exemplo, document, total_paginas):
    print(item["nome"], item["pagina"], item["status"])
```

```python
# CÉLULA 4 — Chamar Gemini (consome API) e validar o resultado
try:
    resultado_analise = analisar_documento(document)
    resultado_validado = validar_topicos(resultado_analise, document, total_paginas)
    for topico in resultado_validado:
        print(f"{topico['nome']} — página {topico['pagina']} — {topico['status']}")
except (ValueError, RuntimeError) as erro:
    print(f"Falha: {erro}")
```
